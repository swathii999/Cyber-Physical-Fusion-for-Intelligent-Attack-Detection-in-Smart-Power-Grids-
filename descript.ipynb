```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
import joblib

# ------------------ Load your training dataset ------------------
# Replace with your actual CSV or DataFrame
df = pd.read_csv("UMD_Cyber_Attacks_Dataset.csv")  # Example file
# Ensure the 'description' column exists
if 'description' not in df.columns:
    raise ValueError("Column 'description' not found in your dataset")

# ------------------ Label Encode the 'description' column ------------------
description_encoder = LabelEncoder()
df['description_encoded'] = description_encoder.fit_transform(df['description'].astype(str))

# ------------------ Save the encoder ------------------
joblib.dump(description_encoder, "description_encoder.joblib")
print("✅ description_encoder.joblib saved successfully")
```
