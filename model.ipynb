```python
import torch
import torch.nn.functional as F
from torch_geometric.nn import GCNConv
from torch_geometric.datasets import Planetoid
from torch_geometric.data import DataLoader
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score, confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load Dataset (e.g., Cora)
dataset = Planetoid(root='data/Cora', name='Cora')
data = dataset[0]
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
data = data.to(device)

# 2. Define GNN Model
class GNN(torch.nn.Module):
    def __init__(self, in_channels, hidden_channels, out_channels):
        super().__init__()
        self.conv1 = GCNConv(in_channels, hidden_channels)
        self.conv2 = GCNConv(hidden_channels, out_channels)
    def forward(self, x, edge_index):
        x = F.relu(self.conv1(x, edge_index))
        x = self.conv2(x, edge_index)
        return x

model = GNN(dataset.num_node_features, 16, dataset.num_classes).to(device)

# 3. Train your model (simplified)
optimizer = torch.optim.Adam(model.parameters(), lr=0.01, weight_decay=5e-4)
for epoch in range(100):
    model.train()
    optimizer.zero_grad()
    out = model(data.x, data.edge_index)
    loss = F.cross_entropy(out[data.train_mask], data.y[data.train_mask])
    loss.backward()
    optimizer.step()

# 4. Evaluate
model.eval()
with torch.no_grad():
    out = model(data.x, data.edge_index)
    logits = out[data.test_mask]
    preds = logits.argmax(dim=1).cpu().numpy()
    labels = data.y[data.test_mask].cpu().numpy()

# 5. Compute Metrics
acc = accuracy_score(labels, preds)
f1 = f1_score(labels, preds, average='weighted')
prec = precision_score(labels, preds, average='weighted')
rec = recall_score(labels, preds, average='weighted')
cm = confusion_matrix(labels, preds)
cr = classification_report(labels, preds)

print(f"Accuracy: {acc:.4f}")
print(f"F1 Score: {f1:.4f}")
print(f"Precision: {prec:.4f}")
print(f"Recall: {rec:.4f}")
print("Classification Report:\n", cr)

# Confusion Matrix Visualization
plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False)
plt.title('GNN - Confusion Matrix')
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.tight_layout()
plt.show()
```

```python
import torch
import torch.nn.functional as F
from torch_geometric.nn import GCNConv
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# Base GCN model
class GCN(torch.nn.Module):
    def __init__(self, in_channels, hidden_channels, out_channels):
        super().__init__()
        self.conv1 = GCNConv(in_channels, hidden_channels)
        self.conv2 = GCNConv(hidden_channels, out_channels)
    def forward(self, x, edge_index):
        x = F.relu(self.conv1(x, edge_index))
        return self.conv2(x, edge_index)

# Boosting-GNN ensemble class
class BoostingGNN:
    def __init__(self, base_gnn_cls, num_estimators=3):
        self.models = []
        self.alphas = []
        self.base_gnn_cls = base_gnn_cls
        self.num_estimators = num_estimators

    def fit(self, data):
        y = data.y.to(device)
        train_mask = data.train_mask
        weights = torch.ones_like(y, dtype=torch.float, device=device)
        weights /= weights.sum()

        for m in range(self.num_estimators):
            model = self.base_gnn_cls().to(device)
            optimizer = torch.optim.Adam(model.parameters(), lr=0.01, weight_decay=5e-4)
            for epoch in range(100):
                model.train()
                optimizer.zero_grad()
                out = model(data.x.to(device), data.edge_index.to(device))[train_mask]
                loss_per_sample = F.cross_entropy(out, y[train_mask], reduction='none')
                loss = (weights[train_mask] * loss_per_sample).sum()
                loss.backward()
                optimizer.step()

            with torch.no_grad():
                preds = model(data.x.to(device), data.edge_index.to(device))[train_mask].argmax(dim=1)
            incorrect = (preds != y[train_mask]).float()
            err = (weights[train_mask] * incorrect).sum()
            alpha = 0.5 * torch.log((1 - err) / (err + 1e-9))
            self.alphas.append(alpha.item())

            weights[train_mask] *= torch.exp(alpha * (incorrect * 2 - 1))
            weights /= weights.sum()

            self.models.append(model)

    def predict(self, data):
        logits_sum = None
        for alpha, model in zip(self.alphas, self.models):
            logits = model(data.x.to(device), data.edge_index.to(device))
            if logits_sum is None:
                logits_sum = alpha * logits
            else:
                logits_sum += alpha * logits
        return logits_sum.argmax(dim=1).cpu().numpy()

# Define dataset-specific base model lambda
base_cls = lambda: GCN(dataset.num_node_features, 16, dataset.num_classes)

ensemble = BoostingGNN(base_cls, num_estimators=3)
ensemble.fit(data)

# Evaluate
full_preds = ensemble.predict(data)
test_mask = data.test_mask
labels = data.y[test_mask].cpu().numpy()
preds = full_preds[test_mask]

print("Accuracy:", accuracy_score(labels, preds))
print("Classification Report:\n", classification_report(labels, preds))

cm = confusion_matrix(labels, preds)
plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False)
plt.title('Boosting-GNN - Confusion Matrix')
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.show()
```
